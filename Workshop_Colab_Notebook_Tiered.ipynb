### Getting a project CSV
In a real workshop, participants would upload their own CSV (`Files > Upload` in the left sidebar, or `from google.colab import files; files.upload()`).

For this notebook, we'll **generate a small, realistic "messy" dataset and save it as a CSV** — so every participant works with the exact same file, and we can practice spotting missing values, duplicates, and wrong data types.


````python
import numpy as np
import pandas as pd

np.random.seed(42)
n = 120

study_hours   = np.round(np.random.normal(4, 1.8, n).clip(0, 10), 1)
sleep_hours   = np.round(np.random.normal(6.5, 1.2, n).clip(3, 10), 1)
attendance    = np.round(np.random.normal(80, 12, n).clip(30, 100), 0)
previous_score = np.round(np.random.normal(65, 10, n).clip(20, 100), 0)
study_group   = np.random.choice(["Morning", "Evening"], size=n)

# Exam score depends (noisily) on the inputs above
exam_score = (
    10 + 6*study_hours + 2*sleep_hours + 0.3*attendance + 0.4*previous_score
    + np.random.normal(0, 6, n)
).clip(0, 100).round(0)

raw = pd.DataFrame({
    "study_hours": study_hours,
    "sleep_hours": sleep_hours,
    "attendance_pct": attendance,
    "previous_score": previous_score,
    "study_group": study_group,
    "exam_score": exam_score
})

# --- Intentionally mess it up a bit, like a real-world CSV ---
messy = raw.copy()
messy.loc[messy.sample(8, random_state=1).index, "sleep_hours"] = np.nan          # missing values
messy.loc[messy.sample(5, random_state=2).index, "attendance_pct"] = np.nan       # missing values
messy = pd.concat([messy, messy.sample(4, random_state=3)], ignore_index=True)    # duplicate rows
messy["attendance_pct"] = messy["attendance_pct"].astype(object)                  # wrong dtype (should be numeric)
messy.loc[messy.sample(3, random_state=4).index, "attendance_pct"] = "unknown"    # bad text value

messy.to_csv("workshop_data_raw.csv", index=False)
print("Saved workshop_data_raw.csv with", len(messy), "rows")
````

### 🟢 How to read the inspection output  *(optional, ~5 min)*
- **`df.info()`** lists every column. Compare **Non-Null Count** with the total number of entries at the top: a smaller count means missing values. **Dtype** shows the type: `float64` = decimal number, `int64` = whole number, `object` = text (or a *mixture*).
- 🚩 Red flag: `attendance_pct` should be numbers, but its dtype is `object`. Something non-numeric is hiding in there.
- **`df.isna().sum()`** counts missing values per column. **`df.duplicated().sum()`** counts rows that are exact copies of an earlier row.

**Predict first:** which columns do you expect to have missing values? Then run the next cell to see the actual rows.

````python
# 🟢 See the problems with your own eyes (raw data, before cleaning)
print("Rows where attendance_pct is the text 'unknown':")
print(df[df["attendance_pct"] == "unknown"])

print("\nFirst few rows where sleep_hours is missing (NaN = 'not a number' = empty):")
print(df[df["sleep_hours"].isna()].head())
````

### Cleaning decisions
- `attendance_pct` should be numeric, but the mixed text ("unknown") forced it into a text (object) column. We'll convert it back to numbers and treat "unknown" as missing.
- We'll **drop duplicate rows** — they'd otherwise double-count the same student.
- We'll **fill missing numeric values with the column median** — a simple, transparent choice for a workshop (in a real project you'd document and justify this more carefully).

*Note each change as you make it — this is good practice for any real analysis.*


````python
df_clean = df.copy()

# Fix data type: force attendance_pct to numeric, turning "unknown" into NaN
df_clean["attendance_pct"] = pd.to_numeric(df_clean["attendance_pct"], errors="coerce")

# Remove duplicate rows
before = len(df_clean)
df_clean = df_clean.drop_duplicates()
print(f"Removed {before - len(df_clean)} duplicate rows")

# Fill missing numeric values with the column median
for col in ["sleep_hours", "attendance_pct"]:
    missing_count = df_clean[col].isna().sum()
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
    print(f"Filled {missing_count} missing values in '{col}' with the median")

df_clean.info()
````

**Result:** a clean, usable table (`df_clean`) and a short log of every change we made, printed above.

### 🟢 How to read each chart  *(optional, ~5 min)*
- **Histogram:** each bar counts how many students fall in a score range. Look for the *shape*: one hump (roughly symmetric), a long tail on one side (skewed), or stray bars far from the rest (possible outliers).
- **Scatter plot:** one dot per student. If the dots drift upward left to right, higher study hours go with higher scores. A shapeless cloud means no clear relationship.
- **Boxplot:** the box holds the middle 50% of scores, the line inside is the median, and the whiskers/dots show the spread and unusual values. Compare the boxes for the two groups: do they overlap a lot?

## 4. Inference and Modelling  *(3:35 – 4:05 pm)*

**Big idea:** we only have a **sample** of students, not every student who exists. We use the sample to learn something about the wider group — but any conclusion comes with **uncertainty**, since a different sample might give slightly different numbers.

### A simple group comparison
Let's compare the average exam score between the two study groups.


````python
group_means = df_clean.groupby("study_group")["exam_score"].agg(["mean", "std", "count"])
group_means
````

The means may differ, but with only ~60 students per group, some of that difference could be due to chance rather than a real effect. A full statistical test (e.g. a t-test) would quantify that uncertainty — worth a mention if time allows.

### Building a prediction model
Now let's use **scikit-learn** to:
1. Split the data into a **training set** (to learn from) and a **test set** (to check performance on unseen data).
2. Fit a simple **linear regression** model.
3. Compare it against a **baseline** (always predicting the average score) — a good model should beat this easily.


### 🟢 Key words for this section  *(optional, ~3 min)*
| Word | Plain-English meaning | Everyday analogy |
|---|---|---|
| **Sample** | The students we actually measured | Tasting one spoonful of soup |
| **Population** | Everyone we'd *like* to say something about | The whole pot |
| **Training set** | Data the model is allowed to learn from | Practice questions with answers |
| **Test set** | Data hidden until the end to check the model | The real exam |
| **Baseline** | A dumb guess (always the average) | "Everyone gets 65" |
| **MAE** (mean absolute error) | Average size of the miss, in score points | "On average I'm off by 5 marks" |
| **Linear regression** | Finds a weighted-sum rule such as `score ≈ a + b·study + c·sleep + ...` | A recipe: how much each ingredient adds |

**Predict first:** will the model beat the baseline? By a lot or a little?

````python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

features = ["study_hours", "sleep_hours", "attendance_pct", "previous_score"]
target = "exam_score"

X = df_clean[features]
y = df_clean[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

print("Training rows:", len(X_train), "| Test rows:", len(X_test))
````

````python
# Baseline: always predict the average training score
baseline_prediction = y_train.mean()
baseline_predictions = [baseline_prediction] * len(y_test)
baseline_mae = mean_absolute_error(y_test, baseline_predictions)

print(f"Baseline (always predict {baseline_prediction:.1f}) — MAE: {baseline_mae:.2f} points")
````

````python
# Fit a linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

model_predictions = model.predict(X_test)
model_mae = mean_absolute_error(y_test, model_predictions)

print(f"Linear regression — MAE: {model_mae:.2f} points")
print(f"Improvement over baseline: {baseline_mae - model_mae:.2f} points")
````

````python
# Plain-English explanation
improvement_pct = (baseline_mae - model_mae) / baseline_mae * 100

print("Plain-English summary:")
print(f"- On average, our baseline guess is off by about {baseline_mae:.1f} points.")
print(f"- Our model's predictions are off by about {model_mae:.1f} points on average.")
print(f"- That's roughly a {improvement_pct:.0f}% improvement over just guessing the average.")
print("- This is measured on the TEST set — data the model did not see while training.")
````

### 🟢 Look at real predictions  *(optional, ~3 min)*

````python
# 🟢 See the model's guesses next to the real answers (first 10 test students)
comparison = pd.DataFrame({
    "actual_score": y_test.values,
    "model_guess": model_predictions.round(0),
    "baseline_guess": round(baseline_prediction, 0),
})
comparison["model_error"] = (comparison["actual_score"] - comparison["model_guess"]).abs()
comparison["baseline_error"] = (comparison["actual_score"] - comparison["baseline_guess"]).abs()
print(comparison.head(10))
print("\nAverage model error on these 10 rows:", round(comparison["model_error"].head(10).mean(), 1))
print("Average baseline error on these 10 rows:", round(comparison["baseline_error"].head(10).mean(), 1))
````

**Result:** a prediction score (Mean Absolute Error) with a plain-English explanation, printed above.

### 🟡 Stretch: test, coefficients and more metrics  *(optional, ~8 min)*

````python
# 🟡 STRETCH 1: is the group difference bigger than chance? (Welch's t-test)
from scipy import stats
g_morning = df_clean.loc[df_clean["study_group"] == "Morning", "exam_score"]
g_evening = df_clean.loc[df_clean["study_group"] == "Evening", "exam_score"]
t_stat, p_value = stats.ttest_ind(g_morning, g_evening, equal_var=False)
print(f"Mean difference (Morning - Evening): {g_morning.mean() - g_evening.mean():.2f} points")
print(f"t = {t_stat:.2f}, p = {p_value:.3f}")
print("A large p-value (roughly > 0.05) means a gap this size is easy to get by chance alone.")

# 🟡 STRETCH 2: what did the model learn? (points gained per +1 unit, others held fixed)
coef_table = pd.Series(model.coef_, index=features).round(2).sort_values(ascending=False)
print("\nModel coefficients (points per +1 unit):")
print(coef_table)
print("Intercept:", round(model.intercept_, 2))

# 🟡 STRETCH 3: more metrics + a train-vs-test comparison (a quick overfitting check)
import numpy as np
from sklearn.metrics import mean_squared_error, r2_score
train_mae = mean_absolute_error(y_train, model.predict(X_train))
rmse = np.sqrt(mean_squared_error(y_test, model_predictions))
print(f"\nTrain MAE: {train_mae:.2f} | Test MAE: {model_mae:.2f}")
print(f"Test RMSE: {rmse:.2f} | Test R²: {r2_score(y_test, model_predictions):.2f}")
print("Note: test error is normally a bit HIGHER than train error. If it's lower, that's luck of a small")
print("test set (only ~30 rows), which is why single-split results wobble. You'll see this in Section 5.")
````

````python
# 🔴 CHALLENGE: compare models fairly with k-fold cross-validation, and put a range on a group difference.
import numpy as np
from sklearn.model_selection import cross_val_score, KFold
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import RandomForestRegressor

def compare_models(X, y, n_splits=5, seed=42):
    models = {
        "baseline (mean)": DummyRegressor(strategy="mean"),
        "linear regression": LinearRegression(),
        "random forest": RandomForestRegressor(n_estimators=200, random_state=seed),
    }
    cv = KFold(n_splits=n_splits, shuffle=True, random_state=seed)
    rows = []
    for name, m in models.items():
        scores = -cross_val_score(m, X, y, cv=cv, scoring="neg_mean_absolute_error")
        rows.append({"model": name, "mean_MAE": scores.mean(), "std_MAE": scores.std()})
    return pd.DataFrame(rows).round(2).sort_values("mean_MAE")

def bootstrap_diff_ci(a, b, n_boot=5000, seed=0):
    rng = np.random.default_rng(seed)
    a, b = np.asarray(a), np.asarray(b)
    diffs = [rng.choice(a, len(a)).mean() - rng.choice(b, len(b)).mean() for _ in range(n_boot)]
    return tuple(float(v) for v in np.percentile(diffs, [2.5, 97.5]).round(2))



results = compare_models(X, y)
print(results if results is not None else "⏳ compare_models not written yet.")

ci = bootstrap_diff_ci(df_clean.loc[df_clean["study_group"] == "Morning", "exam_score"],
                       df_clean.loc[df_clean["study_group"] == "Evening", "exam_score"])
print("95% interval for Morning - Evening:", ci if ci is not None else "⏳ bootstrap_diff_ci not written yet.")
````

````python
# 🔴 CHALLENGE: a reusable end-to-end check for ANY table with a numeric target.
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

def quick_report(df, target, test_size=0.25, seed=42):
    data = df.select_dtypes("number").dropna(subset=[target])
    feats = [c for c in data.columns if c != target]
    data = data.fillna(data.median(numeric_only=True))
    X_tr, X_te, y_tr, y_te = train_test_split(data[feats], data[target], test_size=test_size, random_state=seed)
    base = mean_absolute_error(y_te, [y_tr.mean()] * len(y_te))
    lr = LinearRegression().fit(X_tr, y_tr)
    return {
        "n_rows": len(data),
        "features": feats,
        "baseline_MAE": round(base, 2),
        "model_MAE": round(mean_absolute_error(y_te, lr.predict(X_te)), 2),
    }

report = quick_report(df_clean, "exam_score")
print(report or "⏳ Not written yet. Fill in the TODOs above, then re-run.")
````
